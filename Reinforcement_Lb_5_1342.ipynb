```python
import gymnasium as gym
import random
# Function for Random Agent
def random_agent(environment_name):
    # Create environment
    env = gym.make(environment_name)
    # Reset environment
    state, info = env.reset()
    total_reward = 0
    steps = 0
    print("\nEnvironment:", environment_name)
    print("Initial State:", state)
    # Run until episode ends
    while True:
        # Select a random action
        action = env.action_space.sample()
        # Perform the action
        next_state, reward, terminated, truncated, info = env.step(action)
        # Add reward
        total_reward = total_reward + reward
        steps = steps + 1
        # Display some information
        if steps <= 5:
            print("Step:", steps)
            print("Action:", action)
            print("Reward:", reward)
            print("Next State:", next_state)
        # Stop when episode is finished
        if terminated or truncated:
            break
    # Calculate average reward
    average_reward = total_reward / steps
    print("\nTotal Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", round(average_reward, 2))
    env.close()
# Run Random Agent on the three environments
random_agent("CartPole-v1")
random_agent("MountainCar-v0")
random_agent("FrozenLake-v1")
```
