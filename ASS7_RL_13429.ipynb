```python
n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input("Enter reward: "))
    rewards.append(reward)

gamma = float(input("Enter discount factor: "))

undiscounted = sum(rewards)
discounted = 0

for i in range(n):
    discounted += (gamma ** i) * rewards[i]

print("Rewards:", rewards)
print("Undiscounted Return:", undiscounted)
print("Discounted Return:", discounted)1
```

```python
states = [(0, 0), (0, 1), (0, 2),
          (1, 0), (1, 1), (1, 2)]

actions = ["up", "down", "left", "right"]

goal = (0, 2)
obstacle = (1, 1)

gamma = 0.9


def get_reward(state):
    if state == goal:
        return 10
    elif state == obstacle:
        return -1
    else:
        return -1


def get_next_state(state, action):
    row, col = state

    if action == "up":
        row -= 1
    elif action == "down":
        row += 1
    elif action == "left":
        col -= 1
    elif action == "right":
        col += 1

    next_state = (row, col)

    if next_state not in states or next_state == obstacle:
        return state

    return next_state
```

```python
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

state = tuple(map(int, input("Enter current state (row col): ").split()))
action = input("Enter action: ").lower()

if state in states and action in actions:
    next_state = get_next_state(state, action)

    print("Current State:", state)
    print("Next State:", next_state)
    print("Reward:", get_reward(next_state))
else:
    print("Invalid state or action")
```
