```python
import random
M = int(input("Enter number of machines: "))
machines = []
for i in range(M):
    name = input("Enter machine name: ")
    machines.append(name)
print("\nMachines:", machines)
N = [0] * M
R = [0] * M
Q = [0] * M
epsilon = 0.2
N_play = 10
total_reward = 0
print("\n----- Epsilon-Greedy Strategy -----")
for play in range(1, N_play + 1):
    random_number = random.random()
    if random_number < epsilon:
        choice = random.randint(0, M - 1)
        print("\nPlay", play, ": Exploration")
    else:
        choice = Q.index(max(Q))
        print("\nPlay", play, ": Exploitation")
    reward = float(input("Enter reward: "))
    N[choice] = N[choice] + 1
    R[choice] = R[choice] + reward
    Q[choice] = Q[choice] + (reward - Q[choice]) / N[choice]
    total_reward = total_reward + reward
    print("Updated Q-value:", round(Q[choice], 2))
print("\n FINAL RESULTS")
print("Machine\tN\tR\tQ")
for i in range(M):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(Q[i], 2))
print("\nTotal reward after 10 plays:", total_reward)
```
