```python

gamma = 0.9

# Bellman equations
# V0 = 0.9 * (-1 + 0.9 * V1) + 0.1 * (-1)
# V1 = 0.9 * 100 + 0.1 * (-1 + 0.9 * V0)

# Simplified equations
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1

Q1_right = 100 * 0.9 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

print("Assignment 1 Results")
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))

print("Q(S0, Down) =", round(Q0_down, 2))
print("Q(S0, Right) =", round(Q0_right, 2))
print("Q(S1, Right) =", round(Q1_right, 2))
print("Q(S1, Up) =", round(Q1_up, 2))

print("Check V(S0):", round(0.9 * Q0_down + 0.1 * Q0_right, 2))
print("Check V(S1):", round(0.9 * Q1_right + 0.1 * Q1_up, 2))
```

```python

gamma = 0.9

# Solve the two equations directly
V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

Q0_down = -1 + gamma * V1
Q0_right = -1
Q1_right = 90 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

print("Assignment 1 Results")
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))
print("Q(S0, Down) =", round(Q0_down, 2))
print("Q(S0, Right) =", round(Q0_right, 2))
print("Q(S1, Right) =", round(Q1_right, 2))
print("Q(S1, Up) =", round(Q1_up, 2))

print("Check V(S0) =", round(0.9 * Q0_down + 0.1 * Q0_right, 2))
print("Check V(S1) =", round(0.9 * Q1_right + 0.1 * Q1_up, 2))
```

```python

gamma = 0.9
theta = 0.000001

V0 = 0
V1 = 0
iterations = 0

while True:
    new_V0 = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
    new_V1 = 0.9 * 100 + 0.1 * (-1 + gamma * V0)

    delta0 = abs(new_V0 - V0)
    delta1 = abs(new_V1 - V1)

    delta = max(delta0, delta1)

    V0 = new_V0
    V1 = new_V1
    iterations = iterations + 1

    if delta < theta:
        break

# Calculate action values
Q0_down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q0_right = -1

Q1_right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

print("Assignment 2 Results")
print("Iterations =", iterations)
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))

print("Q(S0, Down) =", round(Q0_down, 2))
print("Q(S0, Right) =", round(Q0_right, 2))
print("Q(S1, Right) =", round(Q1_right, 2))
print("Q(S1, Up) =", round(Q1_up, 2))

print("Policy check S0 =", round(V0, 2), "=", round(Q0_down, 2))
print("Policy check S1 =", round(V1, 2), "=", round(Q1_right, 2))
```
