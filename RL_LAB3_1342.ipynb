```python
import random
# Part A - Machine Initialization
machines = ['A', 'B', 'C']
# Number of plays and epsilon value
N_play = 10
epsilon = 0.2
# Part B - Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
print(" Modified Epsilon-Greedy Multi-Armed Bandit ")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)
# Part C, D and E
for play in range(1, N_play + 1):
    # Generate random number
    r = random.random()
    # Find machine with highest Q-value
    highest_machine = max(Q, key=Q.get)
    # Exploration
    if r < epsilon:
        other_machines = [m for m in machines if m != highest_machine]
        selected_machine = random.choice(other_machines)
        decision = "Exploration"
    # Exploitation
    else:
        selected_machine = highest_machine
        decision = "Exploitation"
    print("\nPlay", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Selected machine:", selected_machine)
    # Get reward from user
    reward = float(input("Enter reward: "))
    # Update N
    N[selected_machine] += 1
    # Store reward
    R[selected_machine].append(reward)
    # Incremental sample-average update
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )
    print("Updated Q-value:", round(Q[selected_machine], 2))
# Part F - Display final results
print("\n FINAL RESULTS ")
print("\nParticipation Count (N):")
for machine in machines:
    print(machine, ":", N[machine])
print("\nRewards (R):")
for machine in machines:
    print(machine, ":", R[machine])
print("\nEstimated Q-values:")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))
# Calculate total reward
total_reward = sum(sum(R[machine]) for machine in machines)
print("\nTotal Reward received:", total_reward)
# Best machine
best_machine = max(Q, key=Q.get)
print("Best machine based on Q-value:", best_machine)
```
