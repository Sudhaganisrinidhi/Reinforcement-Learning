```python
gamma = 0.9
theta = 0.001
states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S0", -1),
        "Left": ("S0", -1),
        "Right": ("S1", -1)
    },

    "S1": {
        "Up": ("S1", -1),
        "Down": ("G", 10),
        "Left": ("S0", -1),
        "Right": ("S1", -1)
    }
}

print("STATES:", states)
print("ACTIONS:", actions)
print("Discount factor:", gamma)
print("Normal reward: -1")
print("Wall reward: -1")
print("Goal reward: +10")
print("Convergence threshold:", theta)
print("\nTRANSITION DETAILS")
for state in ["S0", "S1"]:
    for action in actions:
        next_state, reward = transitions[state][action]

        print(state, "--", action, "-->",
              next_state, "Reward =", reward)
V = {
    "S0": 0,
    "S1": 0,
    "G": 0
}
print("\nINITIAL STATE VALUES")
print(V)
iteration = 0
print("\nVALUE ITERATION")
while True:
    iteration += 1
    new_V = V.copy()
    for state in ["S0", "S1"]:
        q_values = []
        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            q_values.append(q)
        new_V[state] = max(q_values)
    delta = max(
        abs(new_V[state] - V[state])
        for state in states
    )
    V = new_V
    print("Iteration:", iteration,
          "Values:", V,
          "Delta:", round(delta, 4))

    if delta < theta:
        break
print("\nFINAL OPTIMAL STATE VALUES")
for state in states:
    print(state, "=", round(V[state], 4))
print("\nACTION VALUES Q*(s, a)")
Q = {}
for state in ["S0", "S1"]:
    Q[state] = {}
    for action in actions:
        next_state, reward = transitions[state][action]
        q = reward + gamma * V[next_state]
        Q[state][action] = q
        print(state, action, "=", round(q, 4))
print("\nOPTIMAL POLICY")
for state in ["S0", "S1"]:
    best_value = max(Q[state].values())
    best_actions = [
        action for action in actions
        if abs(Q[state][action] - best_value) < 1e-9
    ]
    print(state, ":", best_actions)
print("\nBELLMAN OPTIMALITY VERIFICATION")
for state in ["S0", "S1"]:
    max_q = max(Q[state].values())
    print("State:", state)
    print("V*(s) =", round(V[state], 4))
    print("Maximum Q-value =", round(max_q, 4))
    print("Verified:", abs(V[state] - max_q) < 1e-9)
    print()
```
